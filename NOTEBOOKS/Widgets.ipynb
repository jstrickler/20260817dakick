## Interacting with a widget
The easy way to make an interactive widget is to use the `interact()` method. If you don't specify a widget type, it will use an `IntSlider`. 

```python
def f(x=0):
    print(x)

widgets.interact(f, x=(0,10))
```

`interact()` can be used as a decorator, as well

```python
@widgets.interact(x=(0,10))
def f(x=0):
    print(x)
```

## Simple text-entry widget
This widget accepts user input. When the user presses Enter, it calls the callback function, which displays the text. 

```python
text = Text(placeholder="Type here...")  # text input widget

output_area = widgets.Output()

# event handler for input widget
# input_details is dictionary with details about the event
def on_text_change(change):
    with output_area:
        output_area.clear_output()
        print(change['new'])
        # print(change)

text.observe(on_text_change, names="value")  # add handler to widget
display(text, output_area)  # 
```

## Interacting with a graph
We can use a slider to interact with a graph using the `interact()` function

```python
def plot_sine_wave(frequency):
    # Clear the current figure to prevent overlapping lines
    plt.figure(figsize=(8, 4))
    
    # Generate data based on the slider input
    x = np.linspace(0, 2 * np.pi, 500)
    y = np.sin(x * frequency)
    
    # Create the plot
    plt.plot(x, y, color='blue', linewidth=2)
    plt.title(f"Sine Wave with Frequency: {frequency}")
    plt.grid(True)
    plt.ylim(-1.5, 1.5)
    plt.show()

frequency_slider = widgets.FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description='Frequency:')

# 2. Link the function to a FloatSlider widget
interact(
    plot_sine_wave, 
    frequency=frequency_slider
);
```
